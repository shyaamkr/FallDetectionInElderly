```python
import cv2
import numpy as np

height, width = 480, 640
fps, duration = 30, 30

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter('elderly_walk.mp4', fourcc, fps, (width, height))

print("Creating elderly_walk.mp4...")

for frame in range(fps * duration):
    img = np.ones((height, width, 3), dtype=np.uint8) * 240

    cv2.line(img, (0, int(height * 0.85)), (width, int(height * 0.85)), (150, 150, 150), 3)

    y_base = int(height * 0.4)
    walk_phase = frame * 0.4

    head_y = int(y_base + 20 * np.sin(walk_phase * 0.2))
    cv2.circle(img, (int(width * 0.5), head_y), 22, (220, 180, 150), -1)

    torso_start = (int(width * 0.5), int(head_y + 25))
    torso_end = (int(width * 0.52), int(head_y + 80))
    cv2.line(img, torso_start, torso_end, (50, 40, 30), 18)

    arm_l_start = (int(width * 0.45), int(head_y + 40))
    arm_l_end = (
        int(width * 0.35 + 30 * np.sin(walk_phase)),
        int(head_y + 70 + 20 * np.cos(walk_phase))
    )

    arm_r_start = (int(width * 0.55), int(head_y + 40))
    arm_r_end = (
        int(width * 0.65 + 30 * np.sin(walk_phase + np.pi)),
        int(head_y + 70 + 20 * np.cos(walk_phase + np.pi))
    )

    cv2.line(img, arm_l_start, arm_l_end, (60, 50, 40), 12)
    cv2.line(img, arm_r_start, arm_r_end, (60, 50, 40), 12)

    leg_phase = walk_phase * 1.2

    thigh_l_end = (
        int(width * 0.48 + 25 * np.sin(leg_phase)),
        int(torso_end[1] + 45)
    )
    thigh_r_end = (
        int(width * 0.52 + 25 * np.sin(leg_phase + np.pi)),
        int(torso_end[1] + 45)
    )

    calf_l_end = (
        int(thigh_l_end[0] + 20 * np.sin(leg_phase + np.pi / 2)),
        int(thigh_l_end[1] + 35)
    )
    calf_r_end = (
        int(thigh_r_end[0] + 20 * np.sin(leg_phase + np.pi / 2 + np.pi)),
        int(thigh_r_end[1] + 35)
    )

    cv2.line(img, torso_end, thigh_l_end, (40, 30, 25), 14)
    cv2.line(img, torso_end, thigh_r_end, (40, 30, 25), 14)
    cv2.line(img, thigh_l_end, calf_l_end, (40, 30, 25), 12)
    cv2.line(img, thigh_r_end, calf_r_end, (40, 30, 25), 12)

    cv2.ellipse(img, (int(calf_l_end[0]), int(calf_l_end[1])), (15, 8), 0, 0, 180, (30, 25, 20), -1)
    cv2.ellipse(img, (int(calf_r_end[0]), int(calf_r_end[1])), (15, 8), 0, 0, 180, (30, 25, 20), -1)

    cane_start = (int(width * 0.55), int(head_y + 50))
    cane_end = (int(width * 0.7), int(height - 20))
    cv2.line(img, cane_start, cane_end, (139, 69, 19), 8)

    out.write(img)

out.release()
print("Created elderly_walk.mp4 successfully!")
```
